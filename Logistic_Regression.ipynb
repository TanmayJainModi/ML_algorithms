```python
class Logistic_Regression:

  #declaring teh hyperparameters -> learning rate and number of iterations(also called epoch)
  def __init__(self, learning_rate, no_of_iterations):
    self.learning_rate = learning_rate
    self.no_of_iterations = no_of_iterations


  # fitting the model
  def fit(self, X, Y):
    #initializing weight and bias values

    self.X = X
    self.X = np.insert(self.X, 0, 1, axis=1)
    self.w = np.ones(self.X.shape[1])
    self.Y = Y

    #no of datapoints(rows) in the dataset -->m
    #no of features(columns) in the dataset -->n
    self.m , self.n = self.X.shape

    #gradient descent
    for i in range(self.no_of_iterations):
      self.update_weights()


  def update_weights(self):

    #y_hat formula (sigmoid function)
    Y_hat = 1/(1 + np.exp(- (self.X.dot(self.w))))

    #derivatives
    dw = (1/self.m)*np.dot(self.X.T, (self.Y - Y_hat))
    #db = (1/self.m)*np.sum(Y_hat - self.Y)


    #update the weights and bias using gradient descent
    self.w = self.w + self.learning_rate * dw
    #self.b = self.b - self.learning_rate * db

  # Sigmoid equation and decision boundary
  def predict(self,x):
    self.x = x
    self.x = np.insert(self.x, 0, 1, axis=1)
    Y_pred = 1 / (1 + np.exp( - (self.x.dot(self.w)) ))
    Y_pred = np.where( Y_pred > 0.5, 1, 0)
    return Y_pred
```

```python
lr = Logistic_Regression(learning_rate=0.01, no_of_iterations=100000)
lr.fit(X_train, y_train)
```
